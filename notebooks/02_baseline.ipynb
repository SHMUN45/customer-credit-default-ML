# Preprocessing and Baselines

```python
import pandas as pd

df = pd.read_excel("../data/default of credit card clients.xls", header=1)
df.shape
```

```python
df = df.drop(columns="ID")
df = df.drop_duplicates()
df.shape
```

```python
df["EDUCATION"] = df["EDUCATION"].replace({0: 4, 5: 4, 6: 4})
df["MARRIAGE"] = df["MARRIAGE"].replace({0: 3})
```

```python
df["EDUCATION"].value_counts().sort_index()
```

```python
df["MARRIAGE"].value_counts().sort_index()
```

I've now applied my audit decisions to the data. I dropped ID, removed the duplicates and merged the undocumented codes into the "others" catch all.

```python
df = df.rename(columns={"default payment next month": "default"})
```

```python
X = df.drop(columns="default")
y = df["default"]
```

```python
print(X.shape, y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size = 0.2, stratify = y, random_state = 42
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size = 0.25, stratify = y_temp, random_state=42
)
```

```python
print(X_train.shape, X_val.shape, X_test.shape)
```

```python
print(y_train.value_counts(normalize=True))
print(y_val.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```

I have now completed my 60/20/20 split, of 60% train, 20% validation, 20% testing. It's stratified to make sure of a 22% default for each set. To be able to reproduce, i set random_state = 42, and the test set stays untouched until the end.
